Your friend suggests an SVD of the covariance matrix, which is always defined. Here are the results for $U S V^{\top} = \Sigma$,

```python
U, S, Vh = scipy.linalg.svd(Sigma) # returns V' not V
print(f"U =\n {U}")
print(f"S =\n {S}")
print(f"V' =\n {Vh}")
```

Your friend then shows you the following

```python
U_truncated = U[:, :2] # only the first two columns
S_truncated = np.diag(np.sqrt(S[:2])) # first two singular values
A = U_truncated @ S_truncated
print(f"shape(A) = {A.shape}")
print(f"A =\n{A}")
print(f"||A A' - Sigma|| = {norm(A @ A.T - Sigma)}")
```
